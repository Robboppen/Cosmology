```python
import camb
import numpy as np 
import matplotlib.pylab as plt
```

```python
#Fijar parametros cosmologicos

pars=camb.CAMBparams()

pars.set_cosmology(H0=70, ombh2=0.022, omch2=0.122, omk=0)
pars.InitPower.set_params(As=2e-9,ns=0.965, r=0)
pars.set_matter_power(redshifts=[0,1],kmax=10)

results= camb.get_results(pars)
```

```python
results.luminosity_distance
```

```python
z=np.linspace(0,4,100)

DA=results.angular_diameter_distance(z)

plt.plot(z,DA)
plt.show()
```
